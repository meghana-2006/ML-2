```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.0,
    random_state=42
)

# Display the data
plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
# Create K-Means model with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

# Train the model
kmeans.fit(X)

# Get cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

print("Cluster Labels:")
print(labels)

print("\nCluster Centroids:")
print(centroids)
```

```python
# Plot clustered data
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    cmap="viridis",
    s=50
)

# Plot cluster centers
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    color="red",
    marker="X",
    s=200,
    label="Centroids"
)

plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

```python
# Predict cluster for new data points
new_data = np.array([
    [2, 3],
    [-5, -5],
    [8, 4]
])

predictions = kmeans.predict(new_data)

print("New Data Points:")
print(new_data)

print("\nPredicted Cluster:")
print(predictions)

# Visualize predictions
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", s=40)
plt.scatter(
    new_data[:, 0],
    new_data[:, 1],
    color="red",
    marker="*",
    s=250,
    label="New Data Points"
)

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    color="black",
    marker="X",
    s=150,
    label="Centroids"
)

plt.title("K-Means Cluster Prediction")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
