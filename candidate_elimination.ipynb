```python
import pandas as pd

def candidate_elimination(data):
    attributes = list(data.columns[:-1])
    target = data.columns[-1]

    # Most specific hypothesis
    S = ['0'] * len(attributes)

    # Most general hypothesis
    G = [['?'] * len(attributes)]

    for _, row in data.iterrows():
        example = list(row[attributes])
        label = row[target]

        if label == 'Yes':
            # Generalize S
            for i in range(len(S)):
                if S[i] == '0':
                    S[i] = example[i]
                elif S[i] != example[i]:
                    S[i] = '?'

            # Remove G hypotheses inconsistent with S
            G = [g for g in G if all(
                g[i] == '?' or g[i] == S[i]
                for i in range(len(S))
            )]

        else:
            # Specialize G
            new_G = []

            for g in G:
                for i in range(len(attributes)):
                    if g[i] == '?':
                        if S[i] != '?' and S[i] != example[i]:
                            new_h = g.copy()
                            new_h[i] = S[i]
                            new_G.append(new_h)

            G = new_G

        print("Example:", example, "=>", label)
        print("S =", S)
        print("G =", G)
        print()

    return S, G


# Training data
data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'No'],
    ['Sunny', 'Warm', 'High', 'Weak', 'Yes']
], columns=['Sky', 'Temperature', 'Humidity', 'Wind', 'Play'])

S, G = candidate_elimination(data)

print("Final Specific Boundary:", S)
print("Final General Boundary:", G)
```
