```python
# Find-S Algorithm

def find_s(training_data):
    hypothesis = None

    for example in training_data:
        attributes = example[:-1]   # Input attributes
        target = example[-1]        # Output class

        # Consider only positive examples
        if target == "Yes":
            if hypothesis is None:
                hypothesis = attributes.copy()
            else:
                for i in range(len(hypothesis)):
                    if hypothesis[i] != attributes[i]:
                        hypothesis[i] = "?"

    return hypothesis


# Training data
training_data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"]
]

# Find the most specific hypothesis
hypothesis = find_s(training_data)

print("The most specific hypothesis is:")
print(hypothesis)
```
